```python
import pandas as pd
import numpy as np

# File paths
orders_path = '../data/raw/olist_orders_dataset.csv'
items_path = '../data/raw/olist_order_items_dataset.csv'
customers_path = '../data/raw/olist_customers_dataset.csv'
payments_path = '../data/raw/olist_order_payments_dataset.csv'

# Load raw datasets
df_orders = pd.read_csv(orders_path)
df_items = pd.read_csv(items_path)
df_customers = pd.read_csv(customers_path)
df_payments = pd.read_csv(payments_path)

# Verify dataset dimensions
print(f"Orders Shape: {df_orders.shape}")
print(f"Items Shape: {df_items.shape}")
print(f"Customers Shape: {df_customers.shape}")
print(f"Payments Shape: {df_payments.shape}")
```

```python
# Preview first 5 rows
display(df_orders.head())

# Inspect column data types and missing counts
df_orders.info()
```

```python
# Missing values per column
print("--- Missing Values in Orders ---")
print(df_orders.isnull().sum())

# Distribution of order statuses
print("\n--- Order Status Breakdown ---")
print(df_orders['order_status'].value_counts())
```

```python
# Validate customer granularity (total vs unique customer keys)
print(f"Total Customer Records: {len(df_customers)}")
print(f"Unique Customer IDs: {df_customers['customer_unique_id'].nunique()}")

# Primary key integrity check
duplicate_orders = df_orders['order_id'].duplicated().sum()
print(f"Duplicate Order IDs: {duplicate_orders}")
```
